```python
import nibabel as nib
import numpy as np
import os

def create_region_masks(seg_path, out_dir, region_labels=None):
   
    if region_labels is None:
        region_labels = {
            24: "CSF",
            15: "4th ventricle",
            14: "3rd ventricle",
            51: "right putamen",
            43: "right lateral ventricle",
            4: "left lateral ventricle",
            52: "right pallidum",
            12: "left putamen"
        }
    
    
    seg_img = nib.load(seg_path)
    seg_data = seg_img.get_fdata()
    
    seg_data_int = seg_data.astype(int)
    
    os.makedirs(out_dir, exist_ok=True)
    
    affine = seg_img.affine
    header = seg_img.header
    
    for label, region_name in region_labels.items():
        mask = (seg_data_int == label).astype(np.uint8)
        voxel_count = np.sum(mask)
        print(f"Mask for {region_name} has {voxel_count} voxels")
        
        if voxel_count == 0:
            print(f"Warning: No voxels found for {region_name}. Mask will be empty.")
        mask_img = nib.Nifti1Image(mask, affine, header)
        
        safe_region_name = region_name.replace(' ', '_')
        out_path = os.path.join(out_dir, f"{safe_region_name}_mask.nii.gz")
        nib.save(mask_img, out_path)
        print(f"Saved mask for {region_name} at {out_path}")

```
